```python
import pandas as pd

forecast = pd.read_csv(r"C:\Users\rocky\Desktop\ForeSight\data\processed\forecast_demand_30d.csv", parse_dates=["Date"])
inv = pd.read_csv(r"C:\Users\rocky\Desktop\ForeSight\data\processed\fact_inventory.csv", parse_dates=["Snapshot_Date"])

# inventory has monthly snapshots — you only want the most recent one
latest_date = inv["Snapshot_Date"].max()
latest_inv = inv[inv["Snapshot_Date"] == latest_date].copy()

print(latest_date,latest_inv.shape, forecast.shape)
```

# Confirm SKU overlap between forecast and inventory

```python
missing_in_inv = set(forecast["SKU"]) - set(latest_inv["SKU"])
missing_in_forecast = set(latest_inv["SKU"]) - set(forecast["SKU"])
print("Forecast SKUs missing from inventory:", missing_in_inv)
print("Inventory SKUs missing from forecast:", missing_in_forecast)
```

# Expected demand over each SKU's lead time

```python
lead_times = latest_inv.set_index("SKU")["Lead_Time_Days"]

demand_over_lead_time = {}
for sku, lt in lead_times.items():
    sku_forecast = forecast[forecast["SKU"] == sku].sort_values("Date")
    demand_over_lead_time[sku] = sku_forecast["Forecast_Units"].head(lt).sum()

demand_over_lead_time = pd.Series(demand_over_lead_time, name="Expected_Demand_LeadTime")
print("Lead time range:", lead_times.min(), "-", lead_times.max(), "days")
demand_over_lead_time.head()
```

# Project stock position forward

```python
latest_inv = latest_inv.set_index("SKU")
latest_inv["Expected_Demand_LeadTime"] = demand_over_lead_time
latest_inv["Projected_Stock"] = (
    latest_inv["Current_Stock"] + latest_inv["On_Order"] - latest_inv["Expected_Demand_LeadTime"]
)

print("SKUs with negative projected stock:", (latest_inv["Projected_Stock"] < 0).sum())
latest_inv[["Current_Stock", "On_Order", "Expected_Demand_LeadTime",
            "Projected_Stock", "Safety_Stock", "Reorder_Point"]].head(8)
```

# Risk tier

```python
def risk_tier(row):
    if row["Projected_Stock"] <= row["Safety_Stock"]:
        return "High Risk - Stockout"
    elif row["Projected_Stock"] <= row["Reorder_Point"]:
        return "Reorder Soon"
    elif row["Projected_Stock"] > row["Reorder_Point"] * 2:
        return "Overstock Risk"
    else:
        return "Healthy"

latest_inv["Risk_Tier"] = latest_inv.apply(risk_tier, axis=1)
latest_inv["Risk_Tier"].value_counts()
```

```python
# NOTE: 27/50 SKUs land in Overstock Risk with Reorder_Point*2 as the cutoff.
# That's over half the catalog — sanity-check a few of these against sales
# velocity before trusting the tier as-is; the *2 multiplier is a judgment
# call, not derived from the data, so it's a reasonable thing to tune.
latest_inv[latest_inv["Risk_Tier"] == "High Risk - Stockout"][
    ["Current_Stock", "On_Order", "Expected_Demand_LeadTime", "Projected_Stock", "Safety_Stock"]
]
```

# Continuous risk score (days of stock remaining

```python
avg_daily_forecast = forecast.groupby("SKU")["Forecast_Units"].mean()
latest_inv["Avg_Daily_Forecast"] = avg_daily_forecast
latest_inv["Days_Of_Stock_Remaining"] = latest_inv["Current_Stock"] / latest_inv["Avg_Daily_Forecast"]

latest_inv[["Risk_Tier", "Days_Of_Stock_Remaining"]].sort_values("Days_Of_Stock_Remaining").head(10)
```

# Validate against EDA findings

Check whether the known slow movers (SKU025, SKU011, SKU039 — highest zero-sales-day counts from EDA) show up as Overstock Risk, as expected.

```python
known_slow_movers = ["SKU025", "SKU011", "SKU039"]
latest_inv.loc[latest_inv.index.intersection(known_slow_movers),
               ["Current_Stock", "Avg_Daily_Forecast", "Days_Of_Stock_Remaining", "Risk_Tier"]]
```

```python
out = latest_inv.reset_index()[[
    "SKU", "Product_Name", "Category", "Subcategory",
    "Current_Stock", "On_Order", "Safety_Stock", "Reorder_Point",
    "Expected_Demand_LeadTime", "Projected_Stock",
    "Avg_Daily_Forecast", "Days_Of_Stock_Remaining", "Risk_Tier"
]]

out.to_csv(r"C:\Users\rocky\Desktop\ForeSight\data\processed\risk_scores.csv", index=False)
print("Saved", out.shape, r"to C:\Users\rocky\Desktop\ForeSight\data\processed\risk_scores.csv")
out.sort_values("Days_Of_Stock_Remaining").head(10)
```
