# **Check for the duplicated keys**

```python
dupe_count_sales = sales.duplicated(subset=["Date", "SKU"]).sum()
dupe_count_inv   = inv.duplicated(subset=["Snapshot_Date", "SKU"]).sum()

print(dupe_count_sales, dupe_count_inv)
```

```python
if dupe_count_sales:
    # look at them before deciding — are they true duplicates, or same Date+SKU
    print(sales[sales.duplicated(subset=["Date", "SKU"], keep=False)])
    sales = sales.drop_duplicates(subset=["Date", "SKU"], keep="first")
```

# **Check SKU overlap across files before merging**

```python
sku_ids   = set(sku["SKU"])
sales_ids = set(sales["SKU"])
inv_ids   = set(inv["SKU"])

print("sales SKUs not in sku_master:", sales_ids - sku_ids)
print("inventory SKUs not in sku_master:", inv_ids - sku_ids)
# everything in the left set that's not in the right set so use -
```

```python
print(len(sales_ids - sku_ids))   # 0 — sales SKUs all exist in sku_master
print(len(inv_ids - sku_ids))     # 150 — inventory has 150 orphans
```

```python
orphans = sorted(inv_ids - sku_ids)

inv_excluded = inv[inv["SKU"].isin(orphans)]       # keep these — don't just discard
inv_clean    = inv[~inv["SKU"].isin(orphans)]       # only SKUs sku_master can enrich
```

# **Build the sales fact table**

```python
fact_sales = (
    sales
    .merge(cal, left_on="Date", right_on="date", how="left")
    .merge(sku, on="SKU", how="left")
)
fact_sales = fact_sales.drop(columns=["date"])
```

```python
print(fact_sales.shape)                 # (36550, 23) — same row count as sales, more columns
print(fact_sales.isnull().sum().sum())  # 0 - every row matched both calendar and sku_master
```

# **Build the Inventory fact table**

```python
sku_ids = set(sku["SKU"])
orphans = sorted(set(inv["SKU"]) - sku_ids)

inv_excluded = inv[inv["SKU"].isin(orphans)]     # save separately — don't discard
inv_clean    = inv[~inv["SKU"].isin(orphans)]    # only SKUs sku_master can enrich

fact_inventory = inv_clean.merge(sku, on="SKU", how="left")
```

```python
print(fact_inventory.shape)
print(inv_excluded.shape)
print(fact_inventory.isnull().sum().sum())
```

(1200, 15) — inventory rows for the 50 matched SKUs only

(3600, 8)  — the orphan rows, untouched, saved for review

0 — no NaNs snuck through

# **Log Lines**

```python
log_lines = []

def log(msg):
    print(msg)
    log_lines.append(msg)

log(f"Loaded: sales={sales.shape}, inventory={inv.shape}, sku={sku.shape}, calendar={cal.shape}")
log(f"Duplicate keys: sales={dupe_count_sales}, inventory={dupe_count_inv}")
log(f"Inventory SKUs missing from sku_master: {len(orphans)}")
```

# **Final save**

```python
OUT = Path("Processed")
OUT.mkdir(exist_ok=True)

fact_sales.to_csv(OUT / "fact_sales_daily.csv", index=False)
fact_inventory.to_csv(OUT / "fact_inventory.csv", index=False)
inv_excluded.to_csv(OUT / "excluded_inventory_skus.csv", index=False)
sku.to_csv(OUT / "dim_sku.csv", index=False)
```
